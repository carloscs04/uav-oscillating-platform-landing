# **Camera Exploration**

This notebook is an exploratory device, it is merely to understand the commands of the pi and the cameras.

## **I. Cameras**

### **Camera A. Arducam IMX219**
- **Resolution**: 3280 × 2464 
- **Frame Rate**: 1080p @ 30 FPS, 720p @ 60 FPS, 640×480 @ 90 FPS
- **Connection**: MIPI CSI-2 ribbon cable
- **Software**: libcamera / Picamera2 
- **Focus**: Manual
- **Others**: *Low Distortion*, *Lower Latency*, *Fragile ribbon cable, sensitive to vibration and EMI*

### **Camera B. USB Web Camera Module**
- **Resolution**: 1280 x 740
- **Frame Rate**: 30 FPS at 720p
- **Connection**: USB 2.0 (UVC, plug-and-play)
- **Software**: V4L2 / OpenCV `VideoCapture`
- **Focus**: Fixed
- **Others**: *Higher Latency*, *Flexible USB cable, easy to route*

---

## **II. Calibration Process**
- Calibration provides a file that holds the parameters that make image processing adaptable so that the camera can measure distances and view images properly.
