## **A. Creating a Marker**

A walkthrough of what makes a marker:
* ID
* Grid Size
* White Margin (Quiet Zone)

```python
#generate an ArUCO marker

# GRID PARAMETER

# Dictionary: grid size + posible IDs
dictionary = cv2.aruco.getPredefinedDictionary(cv2.aruco.DICT_4X4_50)

    # NOTE:
    # the grid size (4x4) can be changed to 5x5, 6x6 or 7x7 (the bigger the grid, the finer the marker)
        # finer markers need to be detected from a closer angle
    # the amount of possible markers can vary: 50, 100, 250 or 1000

# MARKER ID + SIZE

ID_MARKER = 10        # number from 0 to 100
SIZE_PX = 400        # pixel image size

# MARKER GENERATION + SILENT ZONE

marker = cv2.aruco.generateImageMarker(dictionary, ID_MARKER, SIZE_PX) #this generates the marker
# no border is included yet

# here a white margin for better identification is added (silence zone)
marker = cv2.copyMakeBorder(
    marker, 50, 50, 50, 50, cv2.BORDER_CONSTANT, value=255 #white color val.
)

# MARKER SAVING

name = f"marker_id{ID_MARKER}.png" #name to save the image
cv2.imwrite(name, marker) #the line that actually saves the marker
print(f"Marker saved as {name}")

cv2.imshow("ArUco Marker", marker)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## **B. Marker Detection**

* This code uses the camera to observe the marker

```python
# GRID (the DICTIONARY has to be the same as the one that generated the marker) - DICT_4X4_50
parameters = cv2.aruco.DetectorParameters() # establishes HOW to look for the marker
detector = cv2.aruco.ArucoDetector(dictionary, parameters)

camera = cv2.VideoCapture(0)
if not camera.isOpened():
    raise SystemExit("Can't Open Camera!")

while True:
    ok, frame = camera.read()
    if not ok:
        break

    gris = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    esquinas, ids, rechazados = detector.detectMarkers(gris)

    if ids is not None:
        # Dibuja el contorno y el ID de cada marcador encontrado
        cv2.aruco.drawDetectedMarkers(frame, esquinas, ids)

        for esquina, id_marcador in zip(esquinas, ids.flatten()):
            pts = esquina[0]                 # 4 esquinas (x, y) en píxeles
            centro = pts.mean(axis=0).astype(int)
            cv2.circle(frame, tuple(centro), 5, (0, 0, 255), -1)
            print(f"ID {id_marcador} en el centro {tuple(centro)}")

    cv2.imshow("Deteccion ArUco (q para salir)", frame)
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

camera.release()
cv2.destroyAllWindows()
```
