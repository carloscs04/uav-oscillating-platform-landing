```python
#generate an ArUCO marker
import cv2 #OpenCV

# Diccionario: 4x4 bits internos, 50 IDs posibles (0 a 49)
diccionario = cv2.aruco.getPredefinedDictionary(cv2.aruco.DICT_4X4_50)

ID_MARCADOR = 7        # cualquier número entre 0 y 49
TAMANO_PX = 400        # tamaño de la imagen en píxeles

marcador = cv2.aruco.generateImageMarker(diccionario, ID_MARCADOR, TAMANO_PX)

# Agrega un margen blanco (zona de silencio) para que se detecte bien
marcador = cv2.copyMakeBorder(
    marcador, 50, 50, 50, 50, cv2.BORDER_CONSTANT, value=255
)

nombre = f"marcador_id{ID_MARCADOR}.png"
cv2.imwrite(nombre, marcador)
print(f"Marcador guardado como {nombre}")

cv2.imshow("Marcador ArUco", marcador)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
